```python
import pandas as pd

# Load the engineered features from Part 2
df = pd.read_csv("C:/Users/woons/Documents/Python Projects/capstone_part2/part3_machine learning/data/features_traffic.csv")

# 1. Define Congestion Category (data-driven quartiles)
q1, q2, q3 = df["traffic_volume"].quantile([0.25, 0.5, 0.75]).values

def bucket(v):
    if v <= q1:
        return "Low"
    elif v <= q2:
        return "Medium"
    elif v <= q3:
        return "High"
    return "Severe"

df["congestion_category"] = df["traffic_volume"].apply(bucket)

# 2. Define Weather Conditions
SEVERE_WEATHER = ["Thunderstorm", "Squall", "Snow", "Rain"]
LOW_VISIBILITY_WEATHER = ["Fog", "Mist", "Haze", "Smoke"]

# Create is_low_visibility feature if it was not created in Part 2
if "is_low_visibility" not in df.columns:
    df["is_low_visibility"] = df["weather_main"].isin(LOW_VISIBILITY_WEATHER).astype(int)


# Proxy accident-risk label
# High risk = severe/low-visibility weather conditions occurring
# together with High/Severe congestion.

high_congestion = df["congestion_category"].isin(["High", "Severe"])

risky_weather = (
    df["weather_main"].isin(SEVERE_WEATHER)
    | (df["is_low_visibility"] == 1)
)

df["high_risk"] = (high_congestion & risky_weather).astype(int)

print("Distribution of Proxy Accident-Risk Label:")
print(df["high_risk"].value_counts(normalize=True) * 100)

df.to_csv("C:/Users/woons/Documents/Python Projects/capstone_part2/part3_machine learning/data/features_traffic.csv", index=False)
```
